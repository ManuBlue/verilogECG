```python
import numpy as np

def extend_sample(sample, target_length):
    # Remove padding (trailing zeros)
    sample = list(map(int, sample.split()))
    # sample = [x for x in sample if x != 0]
    # sample = sample[:-40]

    #scale sample from 0 to 4095
    maxVal = max(sample)
    minVal = min(sample)
    sample = [int((x-minVal)/(maxVal-minVal)*4095) for x in sample]

    # Calculate how many times to repeat the sample
    repeat_count = (target_length // len(sample)) + 1
    
    # Concatenate the sample with itself
    extended_sample = sample * repeat_count
    
    # Truncate to the target length
    extended_sample = extended_sample[:target_length]
    
    return extended_sample

# Define the target length (e.g., 187)
target_length = 187 * 1

# Extend each sample
extended_samples = [extend_sample(myStr, target_length) for myStr in strs]
import matplotlib.pyplot as plt

# Plot each extended sample in a different graph
for i, sample in enumerate(extended_samples):
    plt.figure(figsize=(10, 6))
    plt.plot(sample[0:300])
    plt.xlabel('Time')
    plt.ylabel('Amplitude')
    plt.title(f'Extended Sample {i}')
    plt.show()
```

```python
import numpy as np
from scipy.signal import resample
import matplotlib.pyplot as plt

def smooth_signal(signal, window_size=5):
    """Smooth the signal using a simple moving average filter."""
    return np.convolve(signal, np.ones(window_size)/window_size, mode='valid')

segments_list = [segment0, segment1, segment2, segment3, segment4]

segmentsDict = {i: x for i, x in enumerate(segments_list)}

def save_segments_as_sv_arrays(segments, original_fs=350, target_fs=125):
    with open("segments.sv", 'w') as file:
        for beat_class, segment in segments.items():
            file.write(f"// Segment for class {beat_class}\n")
            file.write(f"logic [12:0] segment_{beat_class} [{len(segment)}] = '{{\n")
            for i, value in enumerate(segment):
                file.write(f"13'b0_{value:012b}")
                if i < len(segment) - 1:
                    file.write(",")
            file.write("\n};\n\n")

def save_segments_as_cpp_arrays(segments, original_fs=350, target_fs=125):
    with open("segments.cpp", 'w') as file:
        for beat_class, segment in segments.items():
            # Upsample the segment
            num_samples = int(len(segment) * target_fs / original_fs)
            upsampled_segment = resample(segment, num_samples)
            upsampled_segment = np.round(upsampled_segment).astype(int)  # Convert to integers
            
            # Smooth the upsampled segment
            smoothed_segment = smooth_signal(upsampled_segment)
            
            # Scale values between 0 and 4095
            min_val = np.min(smoothed_segment)
            max_val = np.max(smoothed_segment)
            smoothed_segment = 4095 * (smoothed_segment - min_val) / (max_val - min_val)
            smoothed_segment = np.round(smoothed_segment).astype(int)

            file.write(f"// Segment for class {beat_class}\n")
            file.write(f"std::vector<int> segment_{beat_class} = {{\n")
            for i, value in enumerate(smoothed_segment):
                file.write(f"    {value}")
                if i < len(smoothed_segment) - 1:
                    file.write(",")
            file.write("\n};\n\n")

save_segments_as_sv_arrays(segmentsDict, original_fs=125, target_fs=125)
save_segments_as_cpp_arrays(segmentsDict, original_fs=125, target_fs=350)

# Example plot
original_segment = segment4[:300]
num_samples = int(len(original_segment) * 350 / 125)
upsampled_segment = resample(original_segment, num_samples)
upsampled_segment = np.round(upsampled_segment).astype(int)

# Smooth the upsampled segment
smoothed_segment = smooth_signal(upsampled_segment)

# Scale values between 0 and 4095
min_val = np.min(smoothed_segment)
max_val = np.max(smoothed_segment)
smoothed_segment = 4095 * (smoothed_segment - min_val) / (max_val - min_val)
smoothed_segment = np.round(smoothed_segment).astype(int)

plt.figure(figsize=(12, 6))

plt.subplot(3, 1, 1)
plt.plot(original_segment)
plt.title("Original Segment")
plt.xlabel("Sample Index")
plt.ylabel("Amplitude")

plt.subplot(3, 1, 2)
plt.plot(upsampled_segment)
plt.title("Resampled Segment")
plt.xlabel("Sample Index")
plt.ylabel("Amplitude")

plt.subplot(3, 1, 3)
plt.plot(smoothed_segment)
plt.title("Smoothed and Scaled Segment")
plt.xlabel("Sample Index")
plt.ylabel("Amplitude")

plt.tight_layout()
plt.show()
```
