```python
import numpy as np

def preprocess_ecg(samples):
    # Function to find R-peaks (simple threshold-based approach)
    def find_r_peaks(signal, threshold=0.9):
        peaks = []
        for i in range(1, len(signal) - 1):
            if signal[i] > threshold and signal[i] > signal[i - 1] and signal[i] > signal[i + 1]:
                peaks.append(i)
        return peaks

    # Normalize the samples to the range [0, 1] for peak detection
    normalized_samples = (samples - np.min(samples)) / (np.max(samples) - np.min(samples))

    # Find R-peaks
    r_peaks = find_r_peaks(normalized_samples)

    if len(r_peaks) < 2:
        raise ValueError("Not enough R-peaks found in the signal.")

    # Copy samples from the first R-peak to the second R-peak + 40 samples
    start_index = r_peaks[0]
    end_index = r_peaks[1] + 40
    preprocessed_signal = samples[start_index:end_index]

    # Pad until the output size becomes 187
    if len(preprocessed_signal) < 187:
        preprocessed_signal = np.pad(preprocessed_signal, (0, 187 - len(preprocessed_signal)), 'constant')

    return preprocessed_signal

```
